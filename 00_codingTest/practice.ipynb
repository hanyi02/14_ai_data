```python
student= [{'이름': '김민수' , '국어': 85, '영어': 90, '수학': 78}, 
          {'이름': '이영희', '국어': 77, '영어': 55, '수학': 11},
          {'이름': '박철수', '국어': 99, '영어':58 , '수학':17 }]


print("=== 학생 성적 관리 ===")

for i in student:
    i['총점']= i['국어']+i['영어']+i['수학']
    i['평균']= round(i['총점']/3, 2)

    print(f"이름: {i['이름']} 국어: {i['국어']}점, 영어: {i['영어']}점, 수학: {i['수학']}점")
    print(f"총점: {i['총점']}점, 평균:{i['평균']}점 \n ---")
```

```python
def get_average(*args):
    if len(args)==0:
        return 0
    else:
        total=0
        for num in args:
            total+= num
        return round(total/ len(args), 2)
```

```python
def print_gugudan(dan, start=1, end=9):
    if start> end:
        print("잘못된 입력입니다")
        return
    for i in range(start, end+1):
        print(f"{dan}x{i}={dan*i}")
```
