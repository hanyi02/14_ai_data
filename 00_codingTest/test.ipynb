````python
# 문제 2

# 2- 1. 다음 학생들의 성적 정보를 딕셔너리로 저장하시오

# 학생1: 이름=김민수, 국어=85, 영어=90, 수학=78
# 학생2: 이름=이영희, 국어=92, 영어=88, 수학=95
# 학생3: 이름=박철수, 국어=78, 영어=85, 수학=82

students=[{'이름': '김민수', '국어': 85, '영어': 90, '수학':78},
          {'이름': '이영희', '국어': 92, '영어': 88, '수학':95},
          {'이름': '박철수', '국어': 78, '영어': 85, '수학':82}]

# 2-1. 각 학생의 총점과 평균을 계산하여 딕셔너리에 추가하시오

민수_total=students[0]['국어']+students[0]['영어']+students[0]['수학']
영희_total=students[1]['국어']+students[1]['영어']+students[1]['수학']
철수_total=students[2]['국어']+students[2]['영어']+students[2]['수학']

민수_avg=round(민수_total/3,2)
영희_avg=round(영희_total/3,2)
철수_avg=round(철수_total/3,2)



students.append({민수_total, 영희_total, 철수_total})
students.append({민수_avg, 영희_avg, 철수_avg})
print(students)

````

````python
# 문제 4
# 사용자가 입력한 숫자 N까지의 합을 계산하는 프로그램을 작성하세요.

# **요구사항**:

# - 반목문 사용
# - 양수만 입력 가능
# - 계산 과정도 함께 출력

# **실행 예시**:

# ```
# 숫자를 입력하세요: 5
# 1 + 2 + 3 + 4 + 5 = 15
# ```

n = int(input("숫자를 입력하세요: "))

if n>0:
    for i in range(1, n+1):
        result+=i

        if i<n:
            result+=i
            print(f"{i}", end="+")
            

        else:
            print(f"{i}", end="=")
            
            

    print(f"{result}")
else:
    print("양수만 가능함")



````

````python
# 문제 5
# 구구단을 출력하는 함수를 작성하세요.

# **요구사항**

# - 함수 이름: `print_gugudan`
# - 매개변수: `dan`, `start=1`, `end=9`
# - `for`문으로 구구단 출력
# - 잘못된 범위(`start > end`) 입력 시 `"잘못된 범위입니다"` 출력
# - 반환값: 없음 (`print()`로 직접 출력)


def print_gugudan(dan, start, end):
    if start>end:
        print("잘못된 범위입니다.")
    else:
        for i in range(start, end+1):
            print(f"{dan}x{i}={dan*i}")


a=print_gugudan(2, 7, 8)
print(a)
````
