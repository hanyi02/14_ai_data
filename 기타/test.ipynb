```python
a = {1,2,3,4,5}
b = {3,4,5,6,7}
result=set()

# a에만 있는 것
for x in a:
    if x not in b:
        result.add(x)

# b에만 있는 것
for x in b:
    if x not in a:
        result.add(x)

print(result)


result= {j for j in a if j not in b} ^ {j for j in b if j not in a}

print(result)
```

```python
sentences = ["I love Python", "Python is great", "I enjoy coding"]
result=set()

for i in sentences:
    for j in i.split():
        result.add(j)
print(result)

a = {j for i in sentences for j in i.split()}
print(a)
```

```python
numbers = list(range(1, 21))
result=set()

for i in numbers:
    if i%3==0 or i%7==0:
        result.add(i)
print(result)


a={i for i in numbers if i%3==0 or i%7==0}
print(a)
```

```python
pairs = [(1,2),(2,3),(3,4),(1,3),(2,4)]
result=set()


for j, k in pairs:
    if (j+k)%2==0:
        result.add((j, k))

print(result)

a={(j, k) for j, k in pairs if (j+k)%2==0}
print(a)
```
